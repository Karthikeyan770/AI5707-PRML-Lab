## 2. Load Team 16 synthetic dataset
The supplied pickle has keys `x` and `y`.

```python
import pickle
path = DATA / "team16_Synthetic" / "data.pkl"
with open(path, "rb") as f: data = pickle.load(f)
X = np.asarray(data["x"], dtype=float)
y = np.asarray(data["y"]).ravel()
print("Features:", X.shape, "Labels:", y.shape)
print("Classes:", np.unique(y, return_counts=True))
```

## 3. Train-test split and preprocessing
An 80:20 deterministic split is generated using `myRandom()`. Standardization is fit using training data only.

```python
X_train, X_test, y_train, y_test = deterministic_split(X, y, 0.20)
print("Train:", X_train.shape, "Test:", X_test.shape)
```

## 4. GMM from scratch
For every class, a separate full-covariance Gaussian mixture is fitted by Expectation-Maximization (E-step responsibilities, M-step weights/means/covariances). Prediction selects the class with the largest log-likelihood.

```python
mu, sd = standardize_fit(X_train)
A = standardize_transform(X_train, mu, sd)
B = standardize_transform(X_test, mu, sd)
gmm_pred = gmm_predict(A, y_train, B, K=1)
print("GMM accuracy:", f"{accuracy_score(y_test,gmm_pred)*100:.2f}%")
```

## 5. LDA from scratch
LDA estimates class means and priors, pools within-class scatter, and uses the inverse pooled covariance to form linear discriminant scores.

```python
lda = LDAFromScratch().fit(A, y_train)
lda_pred = lda.predict(B)
print("LDA accuracy:", f"{accuracy_score(y_test,lda_pred)*100:.2f}%")
```

## 6. KNN from scratch
For each test sample, Euclidean distances are calculated to training samples. The majority label among the five nearest neighbours is returned.

```python
knn = KNNFromScratch(k=5).fit(A, y_train)
knn_pred = knn.predict(B)
print("KNN accuracy:", f"{accuracy_score(y_test,knn_pred)*100:.2f}%")
```

## 7. SVM from scratch
The SVM uses an RBF kernel and simplified Sequential Minimal Optimization (SMO). For multiclass problems, one-vs-rest classifiers are trained. A deterministic training subset is used for the kernel model when the dataset is large.

```python
if len(A) > 1200:
    ix = deterministic_permutation(len(A))[:1200]
    A_svm, y_svm = A[ix], y_train[ix]
else:
    A_svm, y_svm = A, y_train
svm = SVMFromScratch(C=2.0, max_passes=4, max_iter=300).fit(A_svm, y_svm)
svm_pred = svm.predict(B)
print("SVM accuracy:", f"{accuracy_score(y_test,svm_pred)*100:.2f}%")
```

## 8. PCA and repeat experiments
PCA is implemented using the covariance matrix and `numpy.linalg.eigh`. The PCA mapping is fit on training features only. Classification is repeated for top K dimensions.

```python
results=[]
for k in [None, 2, 4, 8, 12]:
    if k is not None and k >= A.shape[1]: continue
    scores = evaluate_classifiers(X_train,y_train,X_test,y_test,pca_k=k)
    row={"Features":"Original" if k is None else f"PCA-{k}", **scores}
    results.append(row)
    print(row)
result_df=pd.DataFrame(results)
display(result_df)
ax=result_df.set_index("Features")[["GMM","LDA","KNN","SVM"]].plot(kind="bar",ylim=(0,1),figsize=(10,5))
ax.set_ylabel("Accuracy"); ax.set_title("Synthetic dataset: classifiers before and after PCA"); plt.xticks(rotation=0); plt.tight_layout(); plt.show()
result_df.to_csv(OUT/"synthetic_results.csv",index=False)
```
