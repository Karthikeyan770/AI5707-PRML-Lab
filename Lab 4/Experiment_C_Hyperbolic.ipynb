## Experiment C: Same Mean, Different Covariance

**Objective:** Demonstrate that two Gaussian classes with the same mean but different covariance matrices can produce a hyperbolic quadratic decision boundary.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Same mean for both classes
mu1 = np.array([0.0, 2.0])
mu2 = np.array([0.0, 2.0])

# Different covariance matrices chosen to give an indefinite
# quadratic term and hence a hyperbolic decision boundary.
Sigma1 = np.array([[4.0, 0.0],
                   [0.0, 1.0]])

Sigma2 = np.array([[1.0, 0.0],
                   [0.0, 2.0]])

N = 1000
X1 = np.random.multivariate_normal(mu1, Sigma1, N)
X2 = np.random.multivariate_normal(mu2, Sigma2, N)

print('Mean 1:', mu1)
print('Mean 2:', mu2)
print('Covariance 1:\n', Sigma1)
print('Covariance 2:\n', Sigma2)
```

```python
# Gaussian discriminant function
def discriminant(X, mu, Sigma):
    invS = np.linalg.inv(Sigma)
    sign, logdet = np.linalg.slogdet(Sigma)
    d = X - mu
    return -0.5 * logdet - 0.5 * np.einsum('...i,ij,...j->...', d, invS, d)

# Grid for plotting the decision boundary
x = np.linspace(-7, 7, 600)
y = np.linspace(-5, 9, 600)
Xg, Yg = np.meshgrid(x, y)
grid = np.stack([Xg.ravel(), Yg.ravel()], axis=1)

g1 = discriminant(grid, mu1, Sigma1).reshape(Xg.shape)
g2 = discriminant(grid, mu2, Sigma2).reshape(Xg.shape)
D = g1 - g2

plt.figure(figsize=(9, 7))
plt.scatter(X1[:, 0], X1[:, 1], s=12, alpha=0.35, label='Class 1')
plt.scatter(X2[:, 0], X2[:, 1], s=12, alpha=0.35, label='Class 2')

# D = 0 is the decision boundary
plt.contour(Xg, Yg, D, levels=[0], linewidths=2)
plt.scatter(*mu1, marker='x', s=120, linewidths=3, label='Common mean')

plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.title('Experiment C: Same Mean, Different Covariance – Hyperbolic Decision Boundary')
plt.grid(alpha=0.25)
plt.axis('equal')
plt.legend()
plt.show()
```

## Result and interpretation

Both classes have the same mean, $\mu_1=\mu_2=(0,2)^T$, but different covariance matrices. Since the covariance matrices are unequal, the Gaussian discriminant equation is quadratic. For the selected covariance matrices, the quadratic part is indefinite, so the decision boundary is a **hyperbola**.

For this example, the boundary satisfies

$$\frac{1}{4}x_1^2-\frac{1}{2}(x_2-2)^2=-\frac{1}{2}\ln 2,$$

or equivalently

$$x_1^2-2(x_2-2)^2=-2\ln 2.$$

This is a hyperbola centered at the common mean $(0,2)$.
